```python
import time
import google.generativeai as genai

# --- 1. Configure API Key ---
# Keep your key here:
genai.configure(api_key="GEMINI_API_KEY")

# --- 2. Define the NEW Strict System Instruction with the correct name ---
STRICT_EXPERT_SYSTEM_PROMPT = (
    "You are **'K-Dramabot,'** a friendly, highly knowledgeable, and enthusiastic K-Drama expert. "
    "Your sole function is to provide structured K-drama recommendations based on the user's input. "
    "For every user query, you **MUST** respond with **2 to 5** K-drama recommendations. "
    "Your recommendations **MUST** be for **South Korean Dramas only**. "
    "For each recommendation, you **MUST** include the following four fields in a clear, formatted list or table: "
    "1. **Title** (in bold), 2. **Year**, 3. **Summary** (1-2 sentences), and 4. **Why it matches** the user's request. "
    "Maintain a friendly, enthusiastic, and knowledgeable tone. Use relevant emojis (like 💖, ✨, 📺)."
)

model = genai.GenerativeModel("models/gemini-2.5-flash")

# --- 3. Start a *New* Chat with the updated System Instruction ---
# This resets the chat history, making sure the model knows its new name and persona.
chat = model.start_chat(history=[{
    "role": "user",
    "parts": [STRICT_EXPERT_SYSTEM_PROMPT]
}])

last_request_time = 0

print("✅ K-Dramabot Setup Complete with correct name and persona.")
```

```python
# --- Helper Function to Check for Follow-up ---
def check_for_follow_up():
    """Asks the user if they have any more questions before continuing the loop."""
    while True:
        follow_up_input = input("\nK-Dramabot: Do you have any more questions about these recommendations, or would you like to request a new genre/theme? (Yes/No): ").strip().lower()
        if follow_up_input in ['yes', 'y']:
            # User wants to continue or ask a new question
            return True
        elif follow_up_input in ['no', 'n']:
            # User is done with the recommendations, prompt for next main input
            return False
        else:
            print("Please answer with 'Yes' or 'No'.")

# --- Main Chat Function ---
def run_dramabot_chat():
    """
    Runs the main interactive loop for the K-Drama Recommender chatbot.
    """
    global last_request_time
    REQUEST_DELAY_SECONDS = 5

    # --- System Welcome Message ---
    print("\n--- K-Dramabot Activated ---")
    print("Hello! I'm **K-Dramabot**. I'm ready to find your next favorite K-Drama! ✨")
    print("Tell me the genre or theme you are looking for, and I will give you 2-5 structured recommendations!")
    print("Type 'quit', 'exit', or 'stop' to end the chat.")
    print("----------------------------\n")

    while True:
        try:
            # 1. Get user input
            user_input = input("You: ")

            # Check for exit commands
            if user_input.lower() in ['quit', 'exit', 'stop']:
                print("\nK-Dramabot: Thank you for chatting! Enjoy your next K-Drama binge! 💖")
                break

            # 2. Rate Limit Check
            time_elapsed = time.time() - last_request_time
            if time_elapsed < REQUEST_DELAY_SECONDS:
                wait_time = REQUEST_DELAY_SECONDS - time_elapsed
                print(f" (System: Waiting {wait_time:.2f} seconds to respect rate limit...)")
                time.sleep(wait_time)

            # 3. Send message to Gemini and get response
            response = chat.send_message(user_input)
            last_request_time = time.time()

            # 4. Print the model's response
            print(f"\nK-Dramabot: {response.text}")

            # 5. NEW: Check for follow-up questions
            # We run the check_for_follow_up function. If it returns False, the loop continues
            # and prompts the user immediately for a new genre/query.
            # If it returns True, it loops back to the beginning of the chat loop to get the new 'You:' input.
            if not check_for_follow_up():
                print("\nK-Dramabot: Got it! What new genre or theme are you curious about now? 📺")
                # The 'while True' loop continues here, prompting the user immediately after this message

        except Exception as e:
            print(f"\n[Error] An unexpected error occurred: {e}")
            print("Restarting the loop...")
            time.sleep(REQUEST_DELAY_SECONDS)
            continue

# Start the chatbot
run_dramabot_chat()
```
